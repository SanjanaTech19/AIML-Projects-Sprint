Wine Quality Prediction 

model used - Random Forest Classifier

Importing the dependencies

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
```

Data Collection

```python
wine_dataset = pd.read_csv("wineQualityReds.csv")
```

```python
print(wine_dataset.head())
```

```python
wine_dataset.shape
```

Dropping the "Unnamed: 0" column 

```python
wine_dataset = wine_dataset.drop(columns="Unnamed: 0" , axis=1)
```

Checking for any missing values 

```python
wine_dataset.isnull().sum()
```

Data Analysis and Visualization

```python
wine_dataset.describe()
```

```python
sns.catplot(x="quality", data=wine_dataset , kind='count', palette="pastel", hue="quality" )
plt.show()
```

quality vs volatile acidity

```python
plot = plt.figure(figsize=(5,5))
sns.barplot(x="quality" , y="volatile.acidity", data=wine_dataset , palette="dark", hue="quality")
plt.show()
```

quality vs citric acid

```python
plot = plt.figure(figsize=(5,5))
sns.barplot(x="quality", y="citric.acid" , data = wine_dataset,palette="pastel", hue="quality")
plt.show()
```

Positive Correlation

```python
correlation = wine_dataset.corr()
```

Heatmap for understanding correlation between columns

```python
plt.figure(figsize=(10,10))
sns.heatmap(correlation , cbar=True, square=True , fmt='.1f' , annot=True , annot_kws={'size':8}, cmap='Blues')
plt.show()
```

Data Preprocessing

Separating data and label

```python
x = wine_dataset.drop(columns="quality" , axis=1)
print(x)
```

Label Binarisation using Lambda function 

```python
y = wine_dataset["quality"].apply(lambda y_value: 1 if y_value >= 7 else 0)
print(y)
```

Splitting Data into train & test data

```python
x_train, x_test , y_train , y_test = train_test_split(x , y , test_size = 0.2 , random_state = 3)
```

```python
print(y.shape , y_train.shape , y_test.shape)
```

Model Training using Random Forest Classifier

```python
model = RandomForestClassifier()
```

```python
model.fit(x_train , y_train)
```

Model Evaluation using accuracy score

Accuracy on test data

```python
x_test_prediction = model.predict(x_test)

test_data_accuracy = accuracy_score(x_test_prediction , y_test)

print("Accuracy : ",test_data_accuracy)
```

Building a predictive system 

```python
input_data = (7.3,0.65,0,1.2,0.065,15,21,0.9946,3.39,0.47,10)

# changing input_data into a numpy array

input_df = pd.DataFrame([input_data] , columns=x.columns)

prediction = model.predict(input_df)
print(prediction)

if (prediction==1):
    print("Good quality wine")
else:
    print("Bad quality wine")
```
