Gold Price Prediction

Importing dependencies

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn import metrics
```

Data Collection and Processing

```python
gold_data = pd.read_csv("gld_price_data.csv")
```

```python
gold_data.describe()
```

```python
gold_data["Date"] = gold_data["Date"].str.replace('/','').astype(int)
```

Heatmap

```python
correlation = gold_data.corr()
```

```python
plt.figure(figsize=(8,8))
sns.heatmap(correlation , cbar=True , square=True , fmt='.1f' ,annot=True , annot_kws={'size':8},cmap="Blues")
plt.show()
```

```python
print(correlation['GLD'])
```

Checking the distribution of the gold price

```python
sns.displot(gold_data['GLD'] ,color="Green")
plt.show()
```

Separating features and targets

```python
x = gold_data.drop(columns=["Date","GLD"],axis=1)

y = gold_data['GLD']
```

```python
print(x)
```

```python
print(y)
```

Splitting data into training data and test data

```python
x_train , x_test , y_train , y_test = train_test_split(x, y , test_size = 0.2 , random_state=2)
```

```python
x.shape , x_train.shape , x_test.shape
```

Model Training 

```python
regressor = RandomForestRegressor()
```

```python
regressor.fit(x_train ,y_train)
```

Model Evaluation

```python
x_train_prediction = regressor.predict(x_train)
```

```python
error_score_train = metrics.r2_score(y_train , x_train_prediction)

print("R Squared Error :",error_score_train)
```

```python
x_test_prediction = regressor.predict(x_test)
```

```python
error_score = metrics.r2_score(y_test , x_test_prediction)

print("R Squared Error :",error_score)
```

Comparing Actual values and Predicted values

```python
actual_prices = y_test.values

plt.figure(figsize=(10, 6))

plt.plot(actual_prices, label="Actual Value", color="blue")

plt.plot(x_test_prediction, label="Predicted Value", color="green")

plt.title("Actual vs Predicted Gold Price")
plt.xlabel("No of values")
plt.ylabel("GLD Price")
plt.legend()
plt.show()
```
