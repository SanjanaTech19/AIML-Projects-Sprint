Data Preprocessing

Separating Features & Target

```python
x = parkinsons_data.drop(columns=["name",'status'], axis = 1)

y = parkinsons_data['status']
```

```python
print(x)
```

```python
print(y)
```

Splitting data into training data & test data

```python
x_train , x_test , y_train , y_test = train_test_split(x , y , test_size = 0.2 , random_state=2)
```

```python
x.shape ,x_train.shape , x_test.shape
```

Data Standardization

```python
scaler = StandardScaler()
```

```python
scaler.fit(x_train)
```

```python
x_train = scaler.transform(x_train)
```

```python
x_test = scaler.transform(x_test)
```

```python
print(x_train)
```

Model Training 

```python
model = svm.SVC(kernel='linear')
```

```python
model.fit(x_train ,y_train)
```

Model Evaluation

```python
x_train_prediction = model.predict(x_train)
```

```python
print("Training Data Accuracy :",accuracy_score(y_train , x_train_prediction))
```

```python
x_test_prediction = model.predict(x_test)
```

```python
print("Test Data Accuracy :",accuracy_score(y_test , x_test_prediction))
```

Building  a Predictive System

```python
input_data = (119.99200,157.30200,74.99700,0.00784,0.00007,0.00370,0.00554,0.01109,0.04374,0.42600,0.02182,0.03130,0.02971,0.06545,0.02211,21.03300,0.414783,0.815285,-4.813031,0.266482,2.301442,0.284654)

input_data_as_array = np.array(input_data).reshape(1, -1)

std_data = scaler.transform(input_data_as_array)

prediction = model.predict(std_data)

print(prediction)

if prediction == 0:
    print("The person does not have Parkinson's disease")
else:
    print("The person has Parkinson's disease")
```
